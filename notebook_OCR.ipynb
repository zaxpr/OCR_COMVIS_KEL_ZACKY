# Proyek UAS: Sistem OCR (Optical Character Recognition)
Notebook ini menunjukkan alur end-to-end: load gambar -> preprocessing -> OCR -> evaluasi hasil.

## 1. Import Library

```python
import sys
sys.path.append('src')

import matplotlib.pyplot as plt
import cv2

from preprocessing import preprocess_pipeline, load_image
from ocr_model import run_ocr
from utils import save_text_output, load_ground_truth, evaluate_batch
```

## 2. Load dan Tampilkan Gambar Contoh

```python
image_path = 'data/sample_images/contoh1.jpg'  # ganti sesuai nama file kamu

image = load_image(image_path)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Gambar Asli')
plt.axis('off')
plt.show()
```

## 3. Preprocessing

```python
processed = preprocess_pipeline(image_path, use_deskew=False)

plt.imshow(processed, cmap='gray')
plt.title('Setelah Preprocessing')
plt.axis('off')
plt.show()
```

## 4. Jalankan OCR

```python
hasil_teks = run_ocr(image_path, processed_image=processed, engine='tesseract', lang='ind+eng')
print('=== Hasil OCR ===')
print(hasil_teks)

save_text_output(hasil_teks, 'output.txt')
```

## 5. Evaluasi Akurasi (Opsional)
Membandingkan hasil OCR dengan ground truth pada `data/ground_truth.csv`.

```python
ground_truth = load_ground_truth('data/ground_truth.csv')

# Contoh: kumpulkan prediksi untuk semua file di ground truth
predictions = {}
for filename in ground_truth:
    path = f'data/sample_images/{filename}'
    proc = preprocess_pipeline(path)
    predictions[filename] = run_ocr(path, processed_image=proc, engine='tesseract')

avg_acc, detail = evaluate_batch(predictions, ground_truth)
print(f'Rata-rata akurasi: {avg_acc:.2%}')
detail
```
